# Cyan - Tuning Weight Decay

````python
from pathlib import Path
from ultralytics import YOLO
import torch

# Project folder on Colab or my Mac
cyan_root = (Path("/content") if Path("/content/TrashDetectionDataset").is_dir()
             else Path.home() / "Desktop" / "IAT360_local")
cyan_data = cyan_root / "TrashDetectionDataset"

cyan_settings = dict(
    data=str(cyan_data / "data.yaml"),
    epochs=50, patience=10, imgsz=640, batch=16,
    optimizer="AdamW", lr0=0.001,
    device=0 if torch.cuda.is_available() else (
        "mps" if torch.backends.mps.is_available() else "cpu"
    ),
    workers=0,
    project=str(cyan_root),
)
````

## Test 1 - 0.0001 Weight Decay

````python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

trainingResults = model.train(
    data="/content/TrashDetectionDataset/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    optimizer="AdamW",
    lr0=0.001,
    patience=10,
    weight_decay=0.0001,
    name="cyan_wd_0p0001"
)

validationResults = model.val()
````

## Test 2 - 0.001 Weight Decay

````python
# This run started in Colab. After epoch 42, I resumed it on my Mac.
cyan_model2 = YOLO("yolov8n.pt")
cyan_model2.train(
    **cyan_settings,
    weight_decay=0.001,
    name="Test2_wd_0p001",
)
````

## Test 3 - 0.005 Weight Decay

````python
cyan_model3 = YOLO("yolov8n.pt")
cyan_model3.train(
    **cyan_settings,
    weight_decay=0.005,
    name="Test3_wd_0p005",
)
````

### Validation and Test Results

````python
for run in ["Test1_wd_0p0001", "Test2_wd_0p001", "Test3_wd_0p005"]:
    best = cyan_root / run / "weights" / "best.pt"

    for split, yaml_file in [("val", "data.yaml"), ("test", "group12data.yaml")]:
        scores = YOLO(str(best)).val(
            data=str(cyan_data / yaml_file), split=split,
            imgsz=640, batch=16, device=cyan_settings["device"],
            workers=0, plots=True,
            project=str(cyan_root / "Cyan_evaluation"), name=f"{run}_{split}",
        )
        b = scores.box
        print(f"{run} | {split} | P={b.mp:.4f}, R={b.mr:.4f}, "
              f"mAP50={b.map50:.4f}, mAP50-95={b.map:.4f}")
````
